## Q6. Median value of horsepower

```python
median_before = df['horsepower'].median()
most_frequent = df['horsepower'].mode()[0]
median_after = df['horsepower'].fillna(most_frequent).median()

print(median_before, most_frequent, median_after)
```

## Q7. Sum of weights

```python
df_asia = df.loc[df['origin'] == 'Asia', ['vehicle_weight', 'model_year']]
X = df_asia.head(7).values
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)

print(w.sum())
```
